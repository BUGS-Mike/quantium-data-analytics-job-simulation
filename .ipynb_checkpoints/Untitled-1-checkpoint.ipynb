Import Libraries.....

```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

Load the Data.....

```python
transactions = pd.read_excel("QVI_transaction_data.xlsx")
customers = pd.read_csv("QVI_purchase_behaviour.csv")
```

```python
print(transactions.shape)
print(customers.shape)

transactions.head()
customers.head()
```

basics data checks...

```python
transactions .info()
customers.info()

transactions.describe()
```

```python
transactions . isnull().sum()
customers.isnull().sum()
```

```python
transactions['DATE']= pd.to_datetime('1899-12-30') + pd.to_timedelta(transactions['DATE'], unit='D')
```

```python
transactions= transactions[transactions['PROD_NAME'].str.contains("Salsa", case=False)]
```

```python
transactions['PACK_SIZE']= transactions['PROD_NAME'].str.extract('(\d+)').astype(float)
```

Extract brand name

```python
transactions['BRAND'] = transactions['PROD_NAME'].str.split().str[0]
```

```python
transactions['BRAND'] = transactions['BRAND'].replace({
    "Red": "RRD",
    "Smith": "Smiths",
    "Dorito": "Doritos",
    "Grain": "GrnWves"
})
```

Merge customer data

```python
df = pd.merge(transactions, customers, on="LYLTY_CARD_NBR")
```

```python
df.head()
df.shape
```

Calculate total sales

```python
total_sales = df['TOT_SALES'].sum()
print("Total Sales:", total_sales)
```

Sales by customer lifestage

```python
lifestage_sales = df.groupby("LIFESTAGE")["TOT_SALES"].sum().sort_values(ascending=False)
print(lifestage_sales)
```

Sales by premium customer

```python
premium_sales = df.groupby("PREMIUM_CUSTOMER")["TOT_SALES"].sum()
print(premium_sales)
```

Number of transactions per segment

```python
transactions_segment = df.groupby("LIFESTAGE")["TXN_ID"].count()
print(transactions_segment)
```

Average spending per customer

```python
avg_spend = df.groupby("LYLTY_CARD_NBR")["TOT_SALES"].mean()
print(avg_spend.head())
```

Brand performance

```python
brand_sales = df.groupby("BRAND")["TOT_SALES"].sum().sort_values(ascending=False)
print(brand_sales)
```

Package size popularity

```python
pack_sales = df.groupby("PACK_SIZE")["TOT_SALES"].sum().sort_values(ascending=False)
print(pack_sales)
```

Visualization ...


1.Sales by lifestage

```python
lifestage_sales.plot(kind="bar", color="skyblue")
plt.title("Sales by Customer Lifestage")
plt.ylabel("Total Sales")
plt.show()
```

2.Premium vS budget customers....

```python
premium_sales.plot(kind="bar", color="orange")
plt.title("Sales by Premium Customer Type")
plt.ylabel("Total Sales")
plt.show()
```

3.TOP CHIP BRANDS....

```python
brand_sales.head(10).plot(kind="bar", color="green")
plt.title("Top Chip Brands")
plt.ylabel("Total Sales")
plt.show()
```

4.PACK SIZE PREFERENCE.....


```python
pack_sales.head(10).plot(kind="bar", color="purple")
plt.title("Most Popular Pack Sizes")
plt.ylabel("Total Sales")
plt.show()
```
